```python
def remove_duplicates(s: str) -> str:
    stack = []
    for char in s:
        if stack and stack[-1] == char:
            stack.pop()   # remove duplicate
        else:
            stack.append(char)
    return "".join(stack)

# Example run
print(remove_duplicates("abdddddgfgghfhffdbacabcccccccvmbnvjhgggkfhjfhgdfgdgdgfxgfgdgf"))  # Output: "ca"
```

```python
def all_two_sum(nums, target):
    seen = {}   # value -> list of indices
    results = []

    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            # Add all pairs with this complement
            for idx in seen[complement]:
                results.append([idx, i])
        # Store index for current number
        if num not in seen:
            seen[num] = []
        seen[num].append(i)

    return results

# Example run
print(all_two_sum([2,7,11,15,7], 9))   # Output: [[0,1], [0,4]]
```

```python


import re
aa = open(r'/Users/chugorjicisco.com/Library/CloudStorage/OneDrive-Cisco/Documents/Cisco_scripts/SWAP3/Morogoro_TZ-MG-NCS-560-7-SR02').read()

# 1. Paste your dictionary here
XX = {
    "GigabitEthernet0/2/0/0": "Gi0/0/0/2",
    "GigabitEthernet0/2/0/8": "Gi0/0/0/3",

    "TenGigE0/1/0/16": "Hu0/0/1/1",

    "TenGigE0/4/0/0": "Ten0/0/0/5",
    "TenGigE0/4/0/3": "Ten0/0/0/10",
    "TenGigE0/4/0/4": "Ten0/0/0/4",

    "TenGigE0/5/0/0": "Hu0/1/1/1",
    "TenGigE0/5/0/1": "Hu0/1/1/1",
    "TenGigE0/5/0/2": "Hu0/1/1/1",
    "TenGigE0/5/0/3": "Hu0/1/1/1",
    "TenGigE0/5/0/4": "Hu0/1/1/1",
    "TenGigE0/5/0/5": "Hu0/1/1/1",
    "TenGigE0/5/0/6": "Hu0/0/1/1",
    "TenGigE0/5/0/7": "Hu0/0/1/1",

    "TenGigE0/0/0/16": "Hu0/1/1/1",

    "TenGigE0/4/0/6": "Hu0/1/1/1",
    "TenGigE0/4/0/7": "Hu0/1/1/1",

    "TenGigE0/10/0/1": "Hu0/1/1/1",
    "TenGigE0/10/0/2": "Hu0/1/1/1",
    "TenGigE0/10/0/5": "Hu0/1/1/1",

    "TenGigE0/11/0/2": "Hu0/0/1/1",

    "TenGigE0/10/0/3": "Ten0/1/0/10",
    "TenGigE0/10/0/4": "Ten0/1/0/4",
    "TenGigE0/10/0/6": "Ten0/1/0/2",
    "TenGigE0/10/0/7": "Ten0/1/0/3",

    "TenGigE0/11/0/0": "Hu0/0/1/1",
    "TenGigE0/11/0/1": "Hu0/0/1/1",
    "TenGigE0/11/0/3": "Hu0/0/1/1",

    "TenGigE0/12/0/16": "Ten0/1/0/5",
    "TenGigE0/14/0/16": "Ten0/1/0/6",
    "TenGigE0/15/0/16": "Ten0/0/0/6",

    # LOCAL PORT (excluding NA)
    "GigabitEthernet0/2/0/1": "Gi0/2/0/1",
    "GigabitEthernet0/2/0/2": "Gi0/2/0/2",
    "GigabitEthernet0/2/0/4": "Gi0/2/0/3",
    "GigabitEthernet0/2/0/6": "Gi0/2/0/4",
    "GigabitEthernet0/12/0/0": "Gi0/2/0/5",
    "GigabitEthernet0/15/0/14": "Gi0/2/0/6",

    "TenGigE0/2/0/16": "Hu0/0/2/3",
    "TenGigE0/4/0/5": "Hu0/0/2/3",
    "TenGigE0/11/0/5": "Hu0/0/2/3",
    "TenGigE0/11/0/6": "Hu0/1/1/3",
    "TenGigE0/11/0/7": "Hu0/1/1/3"
}
    

def transform_config(config_text, mapping_dict):
    # This regex identifies common Cisco interface naming patterns
    # (GigabitEthernet, TenGigE, Bundle-Ether, Loopback, etc.)
    iface_regex = r"(?:GigabitEthernet|TenGigE|Bundle-Ether|Loopback|MgmtEth|FortyGigE|HundredGigE)[\d/.]+"
    
    # Sort keys by length (longest first) to ensure sub-interfaces 
    # are matched before main interfaces (e.g., BE13.3 before BE13)
    sorted_keys = sorted(mapping_dict.keys(), key=len, reverse=True)
    bundle = {e.split(' ')[1]: e.split(' ')[4] for e in config_text.splitlines() if 'bundle id' in e}
    active_bundle = {a:b for a,b in bundle.items() if a in sorted_keys}
    output_lines = []
    #print(active_bundle)
    
    for line in config_text.splitlines():
        # Check if the line mentions any interface pattern
        line = line.replace('TZ-MG-NCS-560-7-SR0','TZ-MG-NCS-5504-SR0')
        if re.search(iface_regex, line):
            matched_key = None
            
            # Check if any interface in our dictionary is present in this line
            for key in sorted_keys:
                if key in line or re.search(rf'Bundle-Ether{active_bundle.get(key)}|Loopback|MgmtEth', line):

                    matched_key = key
                    break


            if matched_key:
                # If found in dictionary, swap the name and keep the line
                new_line = line.replace(matched_key, mapping_dict[matched_key])
                output_lines.append(new_line)
            else:
                # If an interface is mentioned but NOT in our dictionary, 
                # we skip this line (effectively removing it)
                output_lines.append("Not Needed")
                continue
        else:
            # If the line doesn't mention an interface at all (e.g., vrf definition, 
            # policy-map, etc.), we keep it.
            output_lines.append(line)
            
    return "\n".join(output_lines)

print(transform_config(aa, XX))
# Optional: Save to a new file
# with open('updated_config.txt', 'w') as f:
#     f.write(updated_config)
```
